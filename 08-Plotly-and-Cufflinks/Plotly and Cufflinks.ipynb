```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 2. Calculate the Spread (Difference between A and B)
df['Spread'] = df['A'] - df['B']

# 3. Create the Subplot figure
# shared_xaxes=True links the zoom/pan across both charts
fig = make_subplots(
    rows=2, cols=1, 
    shared_xaxes=True, 
    vertical_spacing=0.05,     # Brings the charts closer together
    row_heights=[0.7, 0.3]     # Top chart gets 70% space, bottom gets 30%
)

# 4. Add the two original lines to the Top Chart (Row 1)
fig.add_trace(
    go.Scatter(x=df.index, y=df['A'], name='A', mode='lines'),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df.index, y=df['B'], name='B', mode='lines'),
    row=1, col=1
)

# 5. Add the Spread to the Bottom Chart (Row 2)
# We use 'tozeroy' to fill the area under the curve, making it look like the Cufflinks version
fig.add_trace(
    go.Scatter(
        x=df.index, 
        y=df['Spread'], 
        name='Spread', 
        mode='lines',
        fill='tozeroy',        # Fills the spread area
        line=dict(color='rgba(255, 0, 0, 0.5)') # Semi-transparent red
    ),
    row=2, col=1
)
fig.show()
```
